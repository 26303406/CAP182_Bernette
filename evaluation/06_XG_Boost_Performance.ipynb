```python
# Load the XGBoost model and test data

from utils.notebook_imports import *

plt.style.use('seaborn-v0_8-darkgrid'); sns.set_palette('Oranges')

xgb_model = pickle.load(open('xgb_model.pkl', 'rb'))
X_test = pd.read_csv('X_test.csv')
y_test = pd.read_csv('y_test.csv').values.ravel()

X_test_np = np.array(X_test)
y_proba_xgb = xgb_model.predict(X_test_np)
y_pred_xgb = (y_proba_xgb >= 0.5).astype(int)

print('✓ XGBoost Model Loaded & Predictions Complete')
```

## 1. PERFORMANCE METRICS

```python
# Calculate and save performance metrics

metrics = {
    'Accuracy': accuracy_score(y_test, y_pred_xgb),
    'Precision': precision_score(y_test, y_pred_xgb),
    'Recall': recall_score(y_test, y_pred_xgb),
    'F1-Score': f1_score(y_test, y_pred_xgb),
    'ROC-AUC': roc_auc_score(y_test, y_proba_xgb)
}

print('\n' + '='*60)
print('XGBOOST - PERFORMANCE METRICS')
print('='*60)
for metric, value in metrics.items():
    print(f'{metric:15}: {value:.4f}')
print('='*60)

pd.DataFrame(metrics, index=[0]).to_csv('xgb_eval_metrics.csv', index=False)
```

## 2. CONFUSION MATRIX & ERROR ANALYSIS

```python
# Review confusion matrix and error rates

cm_xgb = confusion_matrix(y_test, y_pred_xgb)
tn, fp, fn, tp = cm_xgb.ravel()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

sns.heatmap(cm_xgb, annot=True, fmt='d', cmap='Oranges', ax=ax1, cbar=False, 
            xticklabels=['Inactive', 'Activated'], yticklabels=['Inactive', 'Activated'])
ax1.set_title('Confusion Matrix', fontsize=14, fontweight='bold')
ax1.set_ylabel('True Label'); ax1.set_xlabel('Predicted Label')

fpr_xgb = fp / (fp + tn) if (fp + tn) > 0 else 0
fnr_xgb = fn / (fn + tp) if (fn + tp) > 0 else 0
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0

errors = {'TP': tp, 'TN': tn, 'FP': fp, 'FN': fn, 'FPR': fpr_xgb, 'FNR': fnr_xgb, 'Specificity': specificity, 'Sensitivity': sensitivity}
error_df = pd.DataFrame(errors, index=[0])

error_names = ['Sensitivity', 'Specificity', 'FPR', 'FNR']
error_vals = [sensitivity, specificity, fpr_xgb, fnr_xgb]
colors = ['green' if v >= 0.95 else 'orange' if v >= 0.85 else 'red' for v in error_vals]
ax2.barh(error_names, error_vals, color=colors, alpha=0.7)
ax2.set_xlim([0, 1]); ax2.set_title('Error Rates', fontsize=14, fontweight='bold')
for i, v in enumerate(error_vals):
    ax2.text(v + 0.02, i, f'{v:.2%}', va='center')

plt.tight_layout(); plt.savefig('xgb_eval_confusion_matrix.png', dpi=300, bbox_inches='tight'); plt.show()

print('\n' + '='*60)
print('ERROR ANALYSIS')
print('='*60)
print(error_df.to_string(index=False))
error_df.to_csv('xgb_eval_errors.csv', index=False)
```

## 3. ROC CURVE ANALYSIS

```python
# Plot the ROC curve

fpr, tpr, thresholds = roc_curve(y_test, y_proba_xgb)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(10, 8))
plt.plot(fpr, tpr, color='orange', lw=2.5, label=f'ROC Curve (AUC = {roc_auc:.4f})')
plt.plot([0, 1], [0, 1], 'k--', alpha=0.3, linewidth=1.5, label='Random Classifier')
plt.xlabel('False Positive Rate', fontsize=12); plt.ylabel('True Positive Rate', fontsize=12)
plt.title('XGBoost - ROC Curve', fontsize=14, fontweight='bold')
plt.legend(loc='lower right', fontsize=11); plt.grid(alpha=0.3)
plt.savefig('xgb_eval_roc_curve.png', dpi=300, bbox_inches='tight'); plt.show()

print(f'\nROC-AUC Score: {roc_auc:.4f}')
```

## 4. PROBABILITY DISTRIBUTION

```python
# Compare predicted probability distributions

plt.figure(figsize=(12, 5))
plt.hist(y_proba_xgb[y_test == 0], bins=40, alpha=0.6, label='Inactive (y=0)', color='red', edgecolor='black')
plt.hist(y_proba_xgb[y_test == 1], bins=40, alpha=0.6, label='Activated (y=1)', color='green', edgecolor='black')
plt.axvline(0.5, color='black', linestyle='--', linewidth=2, label='Decision Threshold (0.5)')
plt.xlabel('Predicted Probability', fontsize=12); plt.ylabel('Frequency', fontsize=12)
plt.title('XGBoost - Probability Distribution', fontsize=14, fontweight='bold')
plt.legend(); plt.grid(alpha=0.3)
plt.savefig('xgb_eval_probability_dist.png', dpi=300, bbox_inches='tight'); plt.show()
```

## 5. CLASSIFICATION REPORT

```python
# Display the classification report

print('\n' + '='*60)
print('DETAILED CLASSIFICATION REPORT')
print('='*60)
print(classification_report(y_test, y_pred_xgb, target_names=['Inactive', 'Activated']))
```

## 6. FEATURE IMPORTANCE

```python
# Rank and plot feature importance

feature_importance = pd.DataFrame({
    'Feature': X_test.columns,
    'Importance': xgb_model.feature_importances_
}).sort_values('Importance', ascending=False)

plt.figure(figsize=(12, 8))
top_n = min(15, len(feature_importance))
top_features = feature_importance.head(top_n)
plt.barh(range(len(top_features)), top_features['Importance'].values, color='darkorange', alpha=0.8)
plt.yticks(range(len(top_features)), top_features['Feature'].values)
plt.xlabel('Importance Score', fontsize=12)
plt.title(f'Top {top_n} Feature Importances - XGBoost', fontsize=14, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout(); plt.savefig('xgb_eval_feature_importance.png', dpi=300, bbox_inches='tight'); plt.show()

print('\nTop 10 Features:')
print(feature_importance.head(10).to_string(index=False))
```
